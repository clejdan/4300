## Q10 (4 points). Correct numbers, misleading display

A report compares average wait times at two clinics: 31 minutes at Clinic A and 34 minutes at Clinic B. The bar chart begins at 30 minutes, so Clinic B's bar appears four times taller.

Why is this display misleading even though the two values are correct? Describe or create a better display. Is a truncated axis always inappropriate, or does the answer depend on the context?

**Answer:**

Bar length carries the message in a bar chart, so readers compare bar lengths. With the axis starting at 30, Clinic A's bar is 1 unit tall and Clinic B's is 4 units tall. The eye reads a 4x difference. The real ratio is 34 / 31 = 1.10, so Clinic B's wait is about 10% longer (3 minutes).

A better display starts the bars at 0, so the bar lengths match the real ratio. You can also print the values on the bars, or plot the 3 minute difference directly with a confidence interval.

A truncated axis is not always wrong. It depends on the chart type and the context. A line chart or dot plot encodes values by position, not by length, so a non-zero baseline is acceptable when the axis is clearly labeled, for example a small change in body temperature. A bar chart always needs a zero baseline.

```python
# Left: the misleading version (axis starts at 30). Right: a fair version (axis starts at 0).
clinics = ["Clinic A", "Clinic B"]
waits = [31, 34]

fig, axes = plt.subplots(1, 2, figsize=(9, 4))

axes[0].bar(clinics, waits, color=["#4c72b0", "#dd8452"])
axes[0].set_ylim(30, 35)
axes[0].set_title("Misleading: axis starts at 30")
axes[0].set_ylabel("Average wait time (minutes)")

axes[1].bar(clinics, waits, color=["#4c72b0", "#dd8452"])
axes[1].set_ylim(0, 40)
axes[1].set_title("Fair: axis starts at 0")
axes[1].set_ylabel("Average wait time (minutes)")
for i, w in enumerate(waits):
    axes[1].text(i, w + 0.5, f"{w} min", ha="center")

plt.tight_layout()
plt.show()
```

## Q14 (6 points). What does a missing review rate mean?

Consider two simple rules for missing `reviews_per_month` values: replace all of them with 0, or leave all of them missing. Check `number_of_reviews` and at least one other relevant variable to see whether either rule works for every row.

Based on what you find, write a rule for handling these values and report how many rows are covered by each part of the rule. Compare one summary before and after the change. Did the result behave as expected?

### AI Copilot Activity 2: A cleaning hypothesis

Ask an AI assistant why `reviews_per_month` might be missing and what should be checked before replacing the missing values with 0. Summarize its recommendation, compare it with your results, and explain your final choice.


```python
# Who has a missing reviews_per_month?
rpm_missing = df["reviews_per_month"].isna()
missing_rows = df[rpm_missing]
present_rows = df[~rpm_missing]

print("Rows with missing reviews_per_month:", rpm_missing.sum())
print()
print("Among the missing rows:")
print("  number_of_reviews equal to 0:", (missing_rows["number_of_reviews"] == 0).sum())
print("  number_of_reviews_ltm equal to 0:", (missing_rows["number_of_reviews_ltm"] == 0).sum())
print("  last_review also missing:", missing_rows["last_review"].isna().sum())
print()
print("Among rows where reviews_per_month is present:")
print("  rows with number_of_reviews equal to 0:", (present_rows["number_of_reviews"] == 0).sum())
print()
print("Rows with reviews but a missing rate:", ((df['number_of_reviews'] > 0) & rpm_missing).sum())
```

**AI recommendation:** The AI said a review rate is usually missing because a listing has no reviews yet, so the rate cannot be computed. It recommended checking that `number_of_reviews` is 0 (and `last_review` is blank) before replacing with 0, because a listing with reviews but a missing rate would be a data problem, and 0 would be wrong for it. (Anthropic, Claude, response to "why might reviews_per_month be missing and what should I check before replacing with 0," accessed 9-25-26.)

**Evidence from the data:** 243 rows are missing `reviews_per_month`. All 243 have `number_of_reviews` = 0, `number_of_reviews_ltm` = 0, and a blank `last_review`. None of the 2,609 rows with a rate has 0 total reviews. No row has reviews but a missing rate. So "leave all missing" throws away a meaningful 0, and "replace all with 0" happens to give the right answer for every row in this file.

**Decision:** I use a conditional rule so it stays correct if the data changes. If `reviews_per_month` is missing **and** `number_of_reviews` = 0, set it to 0 (243 rows). If it is missing **and** `number_of_reviews` > 0, leave it missing (0 rows). All other rows keep their values (2,609 rows).

```python
# Conditional rule
before = df["reviews_per_month"].copy()

df["reviews_per_month_clean"] = df["reviews_per_month"]
rule_zero = df["reviews_per_month"].isna() & (df["number_of_reviews"] == 0)
df.loc[rule_zero, "reviews_per_month_clean"] = 0

print("Rows set to 0 by the rule:", rule_zero.sum())
print("Rows left missing (missing but has reviews):",
      (df["reviews_per_month"].isna() & (df["number_of_reviews"] > 0)).sum())
print("Rows unchanged:", (~df["reviews_per_month"].isna()).sum())
print()

summary = pd.DataFrame({
    "before": [before.count(), before.mean(), before.median()],
    "after": [df["reviews_per_month_clean"].count(),
              df["reviews_per_month_clean"].mean(),
              df["reviews_per_month_clean"].median()]
}, index=["non-missing count", "mean", "median"]).round(3)
display(summary)
```

**Did the result behave as expected?** Yes. The non-missing count rises from 2,609 to 2,852 (the 243 added rows). The mean falls from 2.178 to 1.992 and the median from 1.64 to 1.435. This is expected, because the 243 new values are all 0, which pulls both summaries down. The change reflects real inactivity (listings with no reviews), so the lower averages are more honest than the earlier ones, which described only listings that had reviews.

Create `df_price` using listings with a recorded price. Report how many rows were excluded and the corresponding percentage. Use numerical summaries and visualization(s) to decide whether the mean gives a fair picture of a typical listing.

Compare at least two measures of a typical price and show both the full distribution and the range containing most listings. Then write a more informative one-sentence statement for the report. Why are listings with missing prices unusable for this question but not necessarily bad records?


```python
# df_price: listings with a recorded price
df_price = df[df["price"].notna()].copy()
excluded = len(df) - len(df_price)
print("Rows in df:", len(df))
print("Rows in df_price:", len(df_price))
print(f"Rows excluded: {excluded} ({excluded / len(df) * 100:.2f}%)")
print()

p = df_price["price"]
trimmed = p[(p >= p.quantile(0.05)) & (p <= p.quantile(0.95))]
print("Mean:", round(p.mean(), 2))
print("Median:", p.median())
print("5% trimmed mean:", round(trimmed.mean(), 2))
print("Most common price (mode):", p.mode().iloc[0])
print("Std deviation:", round(p.std(), 2))
print("Skewness:", round(p.skew(), 2))
print(f"Share of listings priced below the mean: {(p < p.mean()).mean() * 100:.1f}%")
print()
print(p.quantile([0.05, 0.25, 0.5, 0.75, 0.95]).rename("price quantiles"))

# Why are prices missing? Compare availability
print()
print("Missing-price rows with availability_365 = 0:",
      (df[df['price'].isna()]['availability_365'] == 0).sum(), "of", excluded)
print(f"Share of priced rows with availability_365 = 0: "
      f"{(df_price['availability_365'] == 0).mean() * 100:.1f}%")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(p, bins=60, color="#4c72b0")
axes[0].axvline(p.mean(), color="red", linestyle="--", label=f"Mean ${p.mean():.0f}")
axes[0].axvline(p.median(), color="green", linestyle="-", label=f"Median ${p.median():.0f}")
axes[0].set_title("Full price distribution (all priced listings)")
axes[0].set_xlabel("Price per night ($)")
axes[0].set_ylabel("Number of listings")
axes[0].legend()

low, high = p.quantile(0.05), p.quantile(0.95)
middle = p[(p >= low) & (p <= high)]
axes[1].hist(middle, bins=40, color="#55a868")
axes[1].axvline(p.mean(), color="red", linestyle="--", label=f"Mean ${p.mean():.0f}")
axes[1].axvline(p.median(), color="green", linestyle="-", label=f"Median ${p.median():.0f}")
axes[1].set_title(f"Middle 90% of listings (${low:.0f} to ${high:.0f})")
axes[1].set_xlabel("Price per night ($)")
axes[1].set_ylabel("Number of listings")
axes[1].legend()

plt.tight_layout()
plt.show()
```

**Interpretation:**

`df_price` has 2,537 rows. I excluded 315 rows (11.04%) that have no price.

The mean ($196.38) does not give a fair picture of a typical listing. The distribution is strongly right-skewed (skewness 18.9). A few listings priced in the thousands (maximum $9,999) pull the mean up. About 71% of listings cost less than the mean. The median is $133, the 5% trimmed mean is $158.18, and the most common price is $100. The middle 50% of listings cost $95 to $211, and the middle 90% cost $57 to $428.60 (5th to 95th percentile).

**Proposed report sentence:** "Half of Asheville Airbnb listings with a posted price cost $133 per night or less, and the middle half of them cost between $95 and $211."

**Why missing prices are unusable here but not bad records:** A blank price cannot be averaged or ranked, so those rows cannot answer a question about price. But the rows are not necessarily wrong. Of the 315 missing-price rows, 187 (59%) have `availability_365` = 0, versus 1.0% of the priced rows. Many of these listings are probably not open for booking, so no price is shown. The rows still hold valid host, location, and review information for other questions.

## Q16 (6 points). Should the price outliers be removed?

Suppose the platform removes every price flagged by the 1.5 × IQR rule before publishing a dashboard. Find the flagged listings and inspect at least two other variables that may help explain the high prices. Compare one result before and after the listings are removed.

Would you keep this rule, modify it, or reject it? In your answer, distinguish what the IQR rule can detect from what it cannot tell us. Also discuss which type of listing may be lost and what you would check before deleting a record.


```python
p = df_price["price"]
q1, q3 = p.quantile(0.25), p.quantile(0.75)
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr
print(f"Q1 = {q1}, Q3 = {q3}, IQR = {iqr}")
print(f"Lower fence = {lower_fence}, upper fence = {upper_fence}")

flagged = df_price[(df_price["price"] < lower_fence) | (df_price["price"] > upper_fence)]
kept = df_price.drop(flagged.index)
print(f"Flagged listings: {len(flagged)} ({len(flagged) / len(df_price) * 100:.1f}% of df_price)")
print("Flagged below the lower fence:", (df_price["price"] < lower_fence).sum())
print()

print("Room type share, flagged vs all priced listings:")
print(pd.DataFrame({
    "flagged_count": flagged["room_type"].value_counts(),
    "flagged_pct": (flagged["room_type"].value_counts(normalize=True) * 100).round(1),
    "all_priced_pct": (df_price["room_type"].value_counts(normalize=True) * 100).round(1),
}).fillna(0))
print()

cols = ["minimum_nights", "availability_365", "number_of_reviews", "calculated_host_listings_count"]
compare = pd.DataFrame({
    "flagged_median": flagged[cols].median(),
    "kept_median": kept[cols].median(),
    "flagged_mean": flagged[cols].mean().round(1),
    "kept_mean": kept[cols].mean().round(1),
})
display(compare)

print("Flagged listings with 0 reviews:", (flagged["number_of_reviews"] == 0).sum())
print("Flagged listings with availability_365 = 0:", (flagged["availability_365"] == 0).sum())
print()

before_after = pd.DataFrame({
    "before (all priced)": [len(df_price), df_price["price"].mean(), df_price["price"].median(), df_price["price"].std()],
    "after (IQR removal)": [len(kept), kept["price"].mean(), kept["price"].median(), kept["price"].std()],
}, index=["rows", "mean", "median", "std"]).round(2)
display(before_after)

print("Top 8 flagged listings by price:")
display(flagged.sort_values("price", ascending=False)[
    ["name", "price", "room_type", "availability_365", "number_of_reviews", "calculated_host_listings_count"]
].head(8))
```

**Interpretation:**

The IQR fences are -$79 and $385. No listing falls below the lower fence. The rule flags 173 listings above $385 (6.8% of `df_price`).

**What the flagged listings look like.** 167 of the 173 are entire homes/apartments (96.5%, compared with 90.9% of all priced listings). Their median availability (280 days) and median minimum stay (2 nights) match the kept listings (281 and 2). They have fewer reviews (median 16 vs 63) and belong to hosts with more listings (mean 18.0 vs 9.6 listings per host). So many flagged listings are newer or professionally managed properties, not random errors. Some flagged records do look wrong: three "Kasa" rooms are priced at exactly $9,999, and one listing is named "Test- Do Not Book" ($2,558).

**Before and after.** Removing the flagged listings changes the mean from $196.38 to $147.35, the median from $133 to $126, and the standard deviation from $405.68 to $75.64. The median barely moves. The mean moves a lot.

**Decision: modify the rule.** The IQR rule can detect that a price is unusual compared with the rest of the data. It cannot tell whether the price is an error (a placeholder like 9,999), a real luxury listing, or a different product (a large estate). I would not delete every flagged record. I would remove or flag only records that look like errors, keep the rest, and report the median or results with and without the flagged listings.

**What we would lose:** mostly large, high-end entire homes (cabins, estates, multi-bedroom houses) and listings run by professional multi-property hosts. A dashboard without them would understate what travelers can actually book at the top of the market.

**What I would check before deleting a record:** the listing name and description, the repeated exact values (9,999), the number of bedrooms or guest capacity (not in this dataset), whether the listing has reviews or availability, and how its price compares with similar listings in the same room type and neighbourhood.

## Q17 (5 points). Room type and price

A traveler wants to know which `room_type` is usually cheaper. A host is more interested in which type has the highest possible prices. Use group sizes and a suitable comparison of the price distributions to address both questions without letting a few extreme listings dominate the result.

Do the two questions lead to the same choice of room type? How do unequal group sizes affect the comparison? Identify one important variable that has not been controlled for.


```python
def q90(s):
    return s.quantile(0.90)

room_summary = df_price.groupby("room_type")["price"].agg(
    count="count",
    median="median",
    q25=lambda s: s.quantile(0.25),
    q75=lambda s: s.quantile(0.75),
    p90=q90,
    max="max",
).round(1)
room_summary["share_of_listings_pct"] = (room_summary["count"] / room_summary["count"].sum() * 100).round(1)
display(room_summary.sort_values("median"))

order = room_summary.sort_values("median").index
plt.figure(figsize=(8, 4.5))
sns.boxplot(data=df_price, x="room_type", y="price", order=order, showfliers=True)
plt.yscale("log")
plt.title("Nightly price by room type (log scale)")
plt.xlabel("Room type")
plt.ylabel("Price per night ($, log scale)")
plt.tight_layout()
plt.show()
```

**Interpretation:**

| Room type | Listings | Median | 90th percentile | Max |
|---|---:|---:|---:|---:|
| Shared room | 8 | $39 | $52.50 | $63 |
| Private room | 211 | $71 | $217 | $9,999 |
| Hotel room | 11 | $123 | $368 | $440 |
| Entire home/apt | 2,307 | $139 | $326.40 | $9,999 |

**Traveler (usually cheaper).** Compare medians, because they resist extreme listings. Shared rooms have the lowest median ($39) but only 8 listings. Private rooms are the cheapest type with a meaningful number of listings (median $71, n = 211).

**Host (highest possible prices).** The maximum is not a good measure: the top value for both entire homes and private rooms is the $9,999 value, which looks like a placeholder. I used the 90th percentile instead. Entire homes reach $326.40 at the 90th percentile with 2,307 listings behind it. Hotel rooms are higher ($368) but only 11 listings support that number.

**Same choice?** No. The traveler's best option is a private room (or a shared room, with very few listings), and the host's best option for high prices is an entire home/apt. The two questions ask about different parts of the distribution (the low-middle and the top).

**Unequal group sizes:** Entire homes are 90.9% of the priced listings, and shared rooms are 8 listings. Statistics from tiny groups (shared, hotel) are unstable, so one listing can change them. Boxplots for these groups should be read as hints, not conclusions. Entire homes also dominate the overall average.

**Not controlled for:** size (bedrooms or guest capacity, which the dataset does not contain). A private room and an entire home differ in size, and location (neighbourhood) also affects price.

## Q18 (4 points). Overall and within-group patterns

Consider the claim: **“Listings with more reviews tend to have lower prices.”**

Check whether the pattern looks the same in the full dataset and within the `room_type` groups. You may use a visualization, grouped summaries, or both.

After accounting for room type, is the claim supported, weakened, or more complicated than it first appeared? Give one reason this pattern should not be interpreted as causal.


```python
# Overall pattern
overall_pearson = df_price[["number_of_reviews", "price"]].corr().iloc[0, 1]
overall_spearman = df_price[["number_of_reviews", "price"]].corr(method="spearman").iloc[0, 1]
print(f"Overall: Pearson = {overall_pearson:.3f}, Spearman = {overall_spearman:.3f}")
print()

# Within each room type
rows = []
for room, g in df_price.groupby("room_type"):
    rows.append({
        "room_type": room,
        "n": len(g),
        "pearson": g["number_of_reviews"].corr(g["price"]),
        "spearman": g["number_of_reviews"].corr(g["price"], method="spearman"),
    })
display(pd.DataFrame(rows).round(3))

# Median price by review group (quartiles of number_of_reviews)
d = df_price.copy()
d["review_group"] = pd.qcut(d["number_of_reviews"].rank(method="first"), 4,
                            labels=["Fewest", "Low-mid", "High-mid", "Most"])
print("Median price by review group (all listings):")
print(d.groupby("review_group")["price"].median())
print()
print("Median price by review group and room type:")
display(d.groupby(["room_type", "review_group"])["price"].median().unstack())

# Plot: log-log scatter with room type colors
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_price, x=df_price["number_of_reviews"] + 1, y="price",
                hue="room_type", alpha=0.5)
plt.xscale("log")
plt.yscale("log")
plt.title("Price vs number of reviews, by room type (log scales)")
plt.xlabel("Number of reviews + 1 (log scale)")
plt.ylabel("Price per night ($, log scale)")
plt.tight_layout()
plt.show()
```

**Conclusion about the claim:** The claim is supported, but it is weak. Overall, the Spearman correlation between reviews and price is -0.20 (Pearson -0.10, pulled toward 0 by the extreme prices). The same negative pattern holds inside the two large groups: entire homes -0.22 (n = 2,307) and private rooms -0.22 (n = 211). So the pattern is not just a mix of room types. Median price by review group falls from $147 (fewest reviews) to $112 (most reviews). The quartile medians are not a straight line: entire homes go $154, $165, $143, $116. The small groups disagree or are noisy: hotel rooms are strongly negative (-0.66, n = 11) and shared rooms are positive (+0.25, n = 8), but those groups are too small to trust. A correlation of -0.2 explains only about 4% of the rank variation in price, so "tend to" is the correct wording, and it is a modest tendency.

**Why the pattern is not necessarily causal:** Older listings have had more time to collect reviews, and older or lower-priced listings may differ in other ways (size, quality, location). Lower prices may also attract more bookings, and therefore more reviews. Reviews do not necessarily lower price, and price may cause reviews, or a third factor (listing age, location, capacity) may drive both.

## Q19 (4 points). Looking beyond correlation

Create a correlation table or heatmap containing price and at least four relevant numerical variables. Treat it as a starting point rather than a conclusion.

Choose one relationship that deserves a closer look and one correlation that may be misleading because of skew, grouping, the way a variable is defined, or another data issue. For either pair, examine the data in a second way, for example, with a plot, grouped summary, or selected rows. What does this reveal that the correlation coefficient does not?


```python
num_cols = ["price", "number_of_reviews", "reviews_per_month", "number_of_reviews_ltm",
            "availability_365", "calculated_host_listings_count", "minimum_nights"]

pearson = df_price[num_cols].corr()
spearman = df_price[num_cols].corr(method="spearman")

plt.figure(figsize=(8, 6))
sns.heatmap(pearson, annot=True, fmt=".2f", cmap="coolwarm", center=0, vmin=-1, vmax=1)
plt.title("Pearson correlations (priced listings)")
plt.tight_layout()
plt.show()

print("Price vs host listing count")
print("  Pearson: ", round(pearson.loc["price", "calculated_host_listings_count"], 3))
print("  Spearman:", round(spearman.loc["price", "calculated_host_listings_count"], 3))

no_placeholder = df_price[df_price["price"] < 9999]
print("  Pearson without the three $9,999 listings:",
      round(no_placeholder["price"].corr(no_placeholder["calculated_host_listings_count"]), 3))
print()

# Second look at price vs host listing count: grouped medians
d = df_price.copy()
d["host_group"] = pd.cut(d["calculated_host_listings_count"], [0, 1, 2, 5, 1000],
                         labels=["1", "2", "3-5", "6+"])
display(d.groupby("host_group")["price"].agg(["count", "median", "mean"]).round(1))

plt.figure(figsize=(7, 4.5))
sns.boxplot(data=d, x="host_group", y="price")
plt.yscale("log")
plt.title("Price by number of listings the host has (log scale)")
plt.xlabel("Listings owned by the host")
plt.ylabel("Price per night ($, log scale)")
plt.tight_layout()
plt.show()

print("Availability vs host listing count")
print("  Pearson: ", round(pearson.loc["availability_365", "calculated_host_listings_count"], 3))
print("  Spearman:", round(spearman.loc["availability_365", "calculated_host_listings_count"], 3))
display(d.groupby("host_group")["availability_365"].agg(["count", "median"]))
```

**Interpretation:**

The heatmap shows a few strong pairs, the strongest being `reviews_per_month` and `number_of_reviews_ltm` (0.86). Both measure review activity, so they largely repeat each other. Price has weak correlations with everything else (all between -0.11 and +0.05).

**Misleading correlation: price and `calculated_host_listings_count`.** The Pearson correlation is 0.05, which looks like "no relationship." Extreme prices (three $9,999 listings, a maximum of $9,999 in a skewed column) dominate the Pearson calculation. Without those three listings the Pearson value rises to 0.09, and the Spearman (rank) correlation is 0.22. Grouped medians show the pattern the coefficient hides: the median price rises from $119 (host with 1 listing) to $124 (2), $139 (3 to 5), and $186.50 (6 or more). So a modest but steady increase exists that the near-zero Pearson value hides.

**Relationship to look at more closely: availability and host listing count** (Pearson 0.18). The grouped medians show that hosts with more listings have more days open for booking: the median is 242 days for hosts with 1 listing, 294 for 2, 289 for 3 to 5, and 328 for 6 or more. The coefficient does not show that most of the jump comes between 1 and 2 listings, or that the median gap between the smallest and largest group is about 86 days.

**What the second look reveals:** A coefficient compresses the data into one number. It hides skew, outliers, and group patterns. The grouped medians and box plots show the direction and size of the difference in real units (dollars, days).
