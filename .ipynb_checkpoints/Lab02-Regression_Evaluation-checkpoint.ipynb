```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error

pd.set_option("display.max_columns", 100)
```

## Step 1: Load the same Asheville Airbnb data

Place `Asheville_listings.csv` in the same folder as this notebook.

```python
df = pd.read_csv("Asheville_listings.csv")
print("Shape:", df.shape)
df.head()
```

## Step 2: Prepare the target: why use log(price)?

In Lab 01, you saw that Airbnb prices are **strongly right-skewed**: most listings have moderate prices, while a small number are extremely expensive.

These extreme prices can have a large influence on a regression model and on squared-error measures such as RMSE.

To reduce the influence of extreme values, we model:

$$
\log(1 + \text{price})
$$

instead of raw price.

The log transformation **compresses very large values while preserving the ordering of prices**. This makes the model less dominated by a few extremely expensive listings.

> **Important:** Linear regression does not require us to use log(price).  
> We use it here because of the distribution of this particular target variable.

```python
model_df = df.copy()

# Keep rows with known, positive price.
model_df = model_df[model_df["price"].notna() & (model_df["price"] > 0)].copy()
model_df["log_price"] = np.log1p(model_df["price"])

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(model_df["price"], bins=40)
ax.set_title("Original Airbnb Price Distribution")
ax.set_xlabel("Price ($)")
ax.set_ylabel("Listings")
plt.show()

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(model_df["log_price"], bins=40)
ax.set_title("Log-Transformed Price Distribution")
ax.set_xlabel("log(1 + price)")
ax.set_ylabel("Listings")
plt.show()
```

### Check
1. Why can a few very expensive listings have a large effect on RMSE?
2. Does log-transforming price delete those listings? No
3. What changes: the observations, or the scale of the target?

## Step 3: Create training, validation, and test sets

We use the three sets for different jobs:

- **Training:** learn model coefficients.
- **Validation:** compare models and choose hyperparameters.
- **Test:** evaluate the final selected model once.

We will use approximately 60% training, 20% validation, and 20% test data.

```python
train_df, temp_df = train_test_split(model_df, test_size=0.40, random_state=42)
val_df, test_df = train_test_split(temp_df, test_size=0.50, random_state=42)

print("Training rows:", len(train_df))
print("Validation rows:", len(val_df))
print("Test rows:", len(test_df))
```

> **Important:** From this point forward, do not use the test set to choose features, polynomial degree, alpha, or the model. The test set is reserved for the end.

## Step 4: Start simple：one-feature Linear Regression

Start with `availability_365` as a simple baseline.

```python
def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))

simple_feature = ["availability_365"]

X_train_simple = train_df[simple_feature].fillna(train_df[simple_feature].median())
X_val_simple = val_df[simple_feature].fillna(train_df[simple_feature].median())
y_train = train_df["log_price"]
y_val = val_df["log_price"]

simple_model = LinearRegression()
simple_model.fit(X_train_simple, y_train)

simple_train_rmse = rmse(y_train, simple_model.predict(X_train_simple))
simple_val_rmse = rmse(y_val, simple_model.predict(X_val_simple))

print("Simple Linear Regression")
print("Train RMSE:", round(simple_train_rmse, 3))
print("Validation RMSE:", round(simple_val_rmse, 3))
```

### Check
Is training RMSE alone enough to evaluate this model? Why do we also need validation RMSE?

The two RMSE values are very close. Does that mean this is a good model?

## Step 5: Multiple Linear Regression

Add several numerical characteristics explored in Lab 01.

```python
numeric_features = [
    "minimum_nights",
    "number_of_reviews",
    "reviews_per_month",
    "calculated_host_listings_count",
    "availability_365",
    "number_of_reviews_ltm",
    "latitude",
    "longitude",
]

numeric_preprocess = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

multiple_model = Pipeline([
    ("prep", numeric_preprocess),
    ("model", LinearRegression()),
])

multiple_model.fit(train_df[numeric_features], y_train)
multiple_train_rmse = rmse(y_train, multiple_model.predict(train_df[numeric_features]))
multiple_val_rmse = rmse(y_val, multiple_model.predict(val_df[numeric_features]))

comparison = pd.DataFrame({
    "Model": ["Simple Linear", "Multiple Linear"],
    "Train RMSE": [simple_train_rmse, multiple_train_rmse],
    "Validation RMSE": [simple_val_rmse, multiple_val_rmse],
})
comparison["Gap (Val - Train)"] = comparison["Validation RMSE"] - comparison["Train RMSE"]
comparison.round(3)
```

### Check
Compare the training and validation RMSE. What does the validation–training gap tell you?

## Step 6: Increase model complexity

We now increase model complexity by adding polynomial, interaction, and categorical features.

- Four numerical predictors → degree-3 polynomial and interaction terms
- `room_type` and `neighbourhood` → one-hot encoding

This is still a linear model in its coefficients, but it now has many more derived features.

> We are increasing complexity intentionally for this exercise. More features are not automatically better.

```python
complex_numeric_features = [
    "minimum_nights",
    "number_of_reviews",
    "availability_365",
    "latitude",
]
categorical_features = ["room_type", "neighbourhood"]
features_complex = complex_numeric_features + categorical_features

num_complex = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("poly", PolynomialFeatures(degree=3, include_bias=False)),
    ("scaler", StandardScaler()),
])

cat_complex = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
])

complex_preprocess = ColumnTransformer([
    ("num", num_complex, complex_numeric_features),
    ("cat", cat_complex, categorical_features),
])

complex_linear = Pipeline([
    ("prep", complex_preprocess),
    ("model", LinearRegression()),
])

complex_linear.fit(train_df[features_complex], y_train)
complex_train_rmse = rmse(y_train, complex_linear.predict(train_df[features_complex]))
complex_val_rmse = rmse(y_val, complex_linear.predict(val_df[features_complex]))

X_train_transformed = complex_linear.named_steps["prep"].transform(train_df[features_complex])
print("Original input columns:", len(features_complex))
print("Features after encoding / polynomial expansion:", X_train_transformed.shape[1])
print("High-complexity Linear Train RMSE:", round(complex_train_rmse, 4))
print("High-complexity Linear Validation RMSE:", round(complex_val_rmse, 4))
print("Gap (Val - Train):", round(complex_val_rmse - complex_train_rmse, 4))
```

### Check
Compare this model with the multiple linear regression.

1. Did training RMSE improve?
2. What happened to validation RMSE?
3. Did the validation–training gap become much larger?
4. Why can adding many derived features improve training fit while hurting performance on unseen data?

## Step 7: Ridge Regression：tune alpha with validation data

Ridge adds an L2 penalty to the coefficients. We will compare several values of `alpha` using validation RMSE.

Larger `alpha` = stronger regularization.

```python
ridge_alphas = [0.01, 0.1, 1, 10, 100, 1000]
ridge_rows = []
ridge_models = {}

for alpha in ridge_alphas:
    model = Pipeline([
        ("prep", complex_preprocess),
        ("model", Ridge(alpha=alpha)),
    ])
    model.fit(train_df[features_complex], y_train)
    train_score = rmse(y_train, model.predict(train_df[features_complex]))
    val_score = rmse(y_val, model.predict(val_df[features_complex]))
    ridge_rows.append([alpha, train_score, val_score])
    ridge_models[alpha] = model

ridge_results = pd.DataFrame(ridge_rows, columns=["alpha", "Train RMSE", "Validation RMSE"])
ridge_results.round(4)
```

```python
best_ridge_alpha = ridge_results.loc[ridge_results["Validation RMSE"].idxmin(), "alpha"]
best_ridge = ridge_models[best_ridge_alpha]

print("Chosen Ridge alpha:", best_ridge_alpha)
print("Chosen because it has the lowest VALIDATION RMSE, not because it was fixed in advance.")
```

### Check
As `alpha` increases, what happens to training and validation RMSE? Which `alpha` would you choose, and why?

## Step 8: Lasso Regression： shrink and select

Lasso uses an L1 penalty and can set some coefficients exactly to zero. Again, choose `alpha` using validation RMSE.

```python
lasso_alphas = [0.0001, 0.0003, 0.001, 0.003, 0.01, 0.03, 0.1, 0.3]
lasso_rows = []
lasso_models = {}

for alpha in lasso_alphas:
    model = Pipeline([
        ("prep", complex_preprocess),
        ("model", Lasso(alpha=alpha, max_iter=50000)),
    ])
    model.fit(train_df[features_complex], y_train)
    train_score = rmse(y_train, model.predict(train_df[features_complex]))
    val_score = rmse(y_val, model.predict(val_df[features_complex]))
    nonzero = np.count_nonzero(model.named_steps["model"].coef_)
    lasso_rows.append([alpha, train_score, val_score, nonzero])
    lasso_models[alpha] = model

lasso_results = pd.DataFrame(
    lasso_rows,
    columns=["alpha", "Train RMSE", "Validation RMSE", "Non-zero Coefficients"]
)
lasso_results.round(4)
```

```python
best_lasso_alpha = lasso_results.loc[lasso_results["Validation RMSE"].idxmin(), "alpha"]
best_lasso = lasso_models[best_lasso_alpha]

print("Chosen Lasso alpha:", best_lasso_alpha)
print("Non-zero coefficients:", np.count_nonzero(best_lasso.named_steps["model"].coef_))
```

### Check
1. As Lasso regularization becomes stronger, what happens to the number of non-zero coefficients?
2. Why can we describe Lasso as performing feature selection?
3. Does the sparsest model automatically have the best validation performance?

## Step 9: Compare models using validation data

At this stage we are still allowed to use validation performance to choose the final model. We have **not touched the test set**.

Look for the full story rather than one number: the deliberately high-complexity OLS model may fit training data better but generalize poorly; regularization trades some training fit for better validation behavior.

```python
best_ridge_train = rmse(y_train, best_ridge.predict(train_df[features_complex]))
best_ridge_val = rmse(y_val, best_ridge.predict(val_df[features_complex]))
best_lasso_train = rmse(y_train, best_lasso.predict(train_df[features_complex]))
best_lasso_val = rmse(y_val, best_lasso.predict(val_df[features_complex]))

model_summary = pd.DataFrame({
    "Model": ["Simple Linear", "Multiple Linear", "High-complexity Linear", "Best Ridge", "Best Lasso"],
    "Train RMSE": [simple_train_rmse, multiple_train_rmse, complex_train_rmse, best_ridge_train, best_lasso_train],
    "Validation RMSE": [simple_val_rmse, multiple_val_rmse, complex_val_rmse, best_ridge_val, best_lasso_val],
})
model_summary["Gap (Val - Train)"] = model_summary["Validation RMSE"] - model_summary["Train RMSE"]
model_summary.round(4)
```

### Reading the comparison
A large positive validation–training gap can indicate overfitting. A small gap in either direction is usually not concerning.

Note: A large positive validation–training gap may indicate overfitting. A small gap in either direction is usually not concerning.

## Step 10: Final test evaluation： use the test set once

The code below automatically selects the model with the lowest validation RMSE among the models we compared. Only now do we evaluate it on the held-out test set.

```python
candidate_models = {
    "Simple Linear": (simple_model, simple_feature),
    "Multiple Linear": (multiple_model, numeric_features),
    "High-complexity Linear": (complex_linear, features_complex),
    "Best Ridge": (best_ridge, features_complex),
    "Best Lasso": (best_lasso, features_complex),
}

best_model_name = model_summary.loc[model_summary["Validation RMSE"].idxmin(), "Model"]
final_model, final_features = candidate_models[best_model_name]

# Use the correct target/features for the selected model.
y_test = test_df["log_price"]
X_test_final = test_df[final_features].copy()

if best_model_name == "Simple Linear":
    X_test_final = X_test_final.fillna(train_df[simple_feature].median())

final_test_pred = final_model.predict(X_test_final)
final_test_rmse = rmse(y_test, final_test_pred)

print("Selected model:", best_model_name)
print("Final Test RMSE (log-price scale):", round(final_test_rmse, 4))
```

```python
plt.figure(figsize=(6, 5))

plt.scatter(y_test, final_test_pred, alpha=0.5)

# Perfect prediction line
low = min(y_test.min(), final_test_pred.min())
high = max(y_test.max(), final_test_pred.max())

plt.plot([low, high], [low, high], linestyle="--")

plt.xlabel("Actual log(Price)")
plt.ylabel("Predicted log(Price)")
plt.title("Actual vs. Predicted Log-Prices on the Test Set")

plt.show()
```

### Reading the prediction plot

The dashed line represents perfect predictions. Points farther from the line have larger prediction errors.

**Question:** Look at listings with very high actual prices. Does the model tend to overpredict or underpredict them?

### A single RMSE tells us how large the errors are overall, but a prediction plot can show us where the model systematically struggles.

### Why only once?
If we repeatedly look at test RMSE and then change the model, the test set becomes another validation set. It no longer gives an unbiased final estimate of performance on unseen data.

## Short Demo: Logistic Regression for Classification

The Airbnb analysis above used **linear regression** because price is a continuous target.

Logistic regression is used when the target is a **class**, such as 0/1. This short synthetic example shows the basic workflow:

**features → probability → class**


```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression

# Create a small synthetic binary-classification dataset
X_cls, y_cls = make_classification(
    n_samples=200,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    class_sep=1.2,
    random_state=42
)

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls, y_cls, test_size=0.25, random_state=42, stratify=y_cls
)

log_model = LogisticRegression()
log_model.fit(X_train_cls, y_train_cls)
```

### Probabilities first

`predict_proba()` returns the probability of each class. For a binary problem, column 1 is the probability of **Class 1**.


```python
prob_class1 = log_model.predict_proba(X_test_cls)[:, 1]

prob_class1[:10]
```

### How good are the predicted probabilities?

Before converting probabilities into classes, we can evaluate the probabilities themselves.

**Binary cross-entropy (log loss)** gives a smaller value when the model assigns high probability to the correct class and a larger value when it is confidently wrong.

For this metric, **lower is better**.

```python
from sklearn.metrics import log_loss

bce = log_loss(y_test_cls, prob_class1)
print("Binary Cross-Entropy (Log Loss):", round(bce, 3))
```

For example, if the true class is **1**:

- predicting 0.90 is a good probability prediction
- predicting 0.55 is less certain
- predicting 0.10 is confidently wrong and receives a much larger penalty

The probability matters — not just the final 0/1 decision.

```python
# Same true class (1), but different predicted probabilities
example_probs = [0.90, 0.55, 0.10]

for p in example_probs:
    loss = log_loss([1], [[1 - p, p]], labels=[0, 1])
    print(f"P(Class 1) = {p:.2f}  ->  Log Loss = {loss:.3f}")
```

### Convert probabilities to classes

Using a threshold of **0.5**:

- probability ≥ 0.5 → Class 1
- probability < 0.5 → Class 0


```python
pred_manual = (prob_class1 >= 0.5).astype(int)
pred_sklearn = log_model.predict(X_test_cls)

demo = pd.DataFrame({
    "P(Class 1)": prob_class1[:10],
    "Manual class (threshold=0.5)": pred_manual[:10],
    "sklearn predict()": pred_sklearn[:10],
    "Actual class": y_test_cls[:10]
})

demo.round(3)
```

### Change the threshold

The probabilities do not change when the threshold changes. Only the final class decisions change.


```python
pred_05 = (prob_class1 >= 0.5).astype(int)
pred_07 = (prob_class1 >= 0.7).astype(int)

threshold_demo = pd.DataFrame({
    "P(Class 1)": prob_class1[:12],
    "Class @ 0.5": pred_05[:12],
    "Class @ 0.7": pred_07[:12]
})

threshold_demo.round(3)
```

### Evaluate the final classes

After choosing a threshold, we can also evaluate the predicted classes. A simple starting metric is **accuracy**.

These answer two different questions:

- **Log loss:** How good are the predicted probabilities?
- **Accuracy:** How often are the final class predictions correct?

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test_cls, pred_05)
print("Accuracy (threshold = 0.5):", round(accuracy, 3))
```
